```python
import h5py

H5_PATH = "../outputs/predictor/predictor_compilation_gate_time_1_to_10.h5"

with h5py.File(H5_PATH, "r") as f:
    results = f.get("results")
    if results is None:
        raise ValueError("No 'results' group found in H5 file")

    run_names = list(results.keys())
    print(f"Runs: {len(run_names)}")
    for run_name in run_names[:5]:
        attrs = dict(results[run_name].attrs)
        print(run_name, attrs)
```

```python
import csv
import h5py

CSV_PATH = H5_PATH.replace(".h5", ".csv")

columns = [
    "circuit_name",
    "num_ions",
    "gate_count",
    "success",
    "final_timesteps",
    "cpu_time_seconds",
    "idle_count",
    "gate_time_one_qubit",
    "gate_time_two_qubit",
    "num_pzs",
    "max_ions_per_pz",
    "arch",
]

with h5py.File(H5_PATH, "r") as f:
    results = f.get("results")
    if results is None:
        raise ValueError("No 'results' group found in H5 file")

    with open(CSV_PATH, "w", newline="", encoding="utf-8") as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=columns)
        writer.writeheader()
        for run_name in results.keys():
            attrs = results[run_name].attrs
            row = {key: attrs.get(key) for key in columns}
            writer.writerow(row)

print(f"Wrote {CSV_PATH}")
```
